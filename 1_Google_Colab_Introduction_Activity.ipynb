## 4 - Generate a Code

```python
names = []
for i in range(1, 6):
  names.append(f"name{i}")

print(names)
```

```python
# Assuming you have a list of actual names
actual_names = ["John", "Smith", "Alice", "Bob", "Charlie"]

# Make sure the actual_names list is at least as long as the names list
if len(actual_names) >= len(names):
  formatted_names = [f"{names[i].replace('name', 'Name ')}: {actual_names[i]}" for i in range(len(names))]
  print(", ".join(formatted_names))
else:
  print("Error: The list of actual names is too short.")
```
