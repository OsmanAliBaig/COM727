# Simple Linear Regression Model

**Steps:**

1. Import libraries → Load pandas, sklearn, and numpy.

2. Create dataset → Simple numeric data with one predictor (X1) and target (Y).

3. Define variables → Split into independent (X) and dependent (y).

4. Split data → 70% training, 30% testing.

5. Train model → Fit linear regression on training data.

6. Show coefficients → B₀ = intercept, B₁ = slope.

7. Predict → Use test data to estimate Y values.

8. Compare → Show actual vs predicted values.


## Step 1. Import necessary libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

## Step 2. Create a Dataset

```python
data = {
    'X1': [10, 12, 13, 15, 16, 18, 19, 21, 22, 25],
    'Y':  [200, 230, 250, 270, 280, 310, 320, 340, 350, 370]
}
df = pd.DataFrame(data)
df
```

## Step 3. Define independent (X) and dependent (y) variables

```python
X = df[['X1']]   # independent variable
y = df['Y']      # dependent variable
```

## Step 4. Split data into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## Step 5. Create and train the Linear Regression model by using "X_train and y_train"

```python
model = LinearRegression()
model.fit(X_train, y_train)

# Step 6. Display model coefficients
B0 = model.intercept_
B1 = model.coef_[0]

print(f"Model Equation: Y = {B0:.2f} + {B1:.2f} * X1")
```

## Step 7. Get predictions by using the "X_test" data

```python
y_pred = model.predict(X_test)

# Step 8. Display predictions
results = pd.DataFrame({'X1': X_test['X1'], '   Actual Y': y_test, '   Predicted Y': y_pred.round(0)})
print("\nTest Results:")
print(results)
```
